```python
#ejercicio 3
precio_str = "15990"
descuento_str = "0.15"
cantidad_str = "4"
tiene_iva_str = "True"
precio = int(precio_str)
descuento = float(descuento_str)
cantidad = int(cantidad_str)
tiene_iva = tiene_iva_str == "False"
print(tiene_iva)
iva = 0.21 * precio
desc = descuento * precio
if (tiene_iva) : 
    total = (precio - desc + iva) * cantidad
    print(total)
else:
    total = (precio - desc) * cantidad
    print(total)

```
